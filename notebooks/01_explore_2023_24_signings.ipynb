```python
# Sort new signings by goals and assists

df.sort_values(
    by=["goals", "assists"],
    ascending=False
)[["player", "position", "minutes", "goals", "assists"]]
```

```python
plt.figure(figsize=(10, 6))

plt.bar(df["player"], df["goals"])

plt.title("Goals Scored by Tottenham's 2023/24 New Signings")
plt.xlabel("Player")
plt.ylabel("Premier League Goals")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
df["goals_per_90"] = (
    df["goals"] / df["minutes"] * 90
)

df[[
    "player",
    "minutes",
    "goals",
    "goals_per_90"
]].sort_values(
    "goals_per_90",
    ascending=False
)
```

```python
df["goal_contributions"] = df["goals"] + df["assists"]

df[[
    "player",
    "minutes",
    "goals",
    "assists",
    "goal_contributions"
]].sort_values(
    "goal_contributions",
    ascending=False
)
```
